```python
#Importing libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Reading the dataset and viewing head and tail
df = pd.read_excel('Health Insurance Dataset.xlsx', skiprows=0, nrows=1338)
df
```

#### Parametric (t_statistic) test

```python
# Research question - Does smokers have different BMI than non-smokers?
# H0: There is no difference in BMI between smoker and non-smoker.
# H1: There is a difference in BMI between smoker and non-smoker. 

import scipy.stats as stats

df_smoker_bmi = df.loc[df['smoker'] == 'yes', 'bmi'].tolist()
df_nonsmoker_bmi = df.loc[df['smoker'] == 'no', 'bmi'].tolist()

t_stat, p_value = stats.ttest_ind(df_smoker_bmi, df_nonsmoker_bmi, equal_var=True) #t-test w/equal variance between smokers and non-smokers

print('t_stat', t_stat)
print('p_value', p_value)
print('threshold 0.05')

if p_value < 0.05:
    print("There is a statistically significant difference thus we reject the null hypothesis.")
else:
    print("There is no statistically significant difference thus we fail to reject the null hypothesis")
    

#since p-value came much greater than the statistical significance threshold of 0.05, then there is no difference in BMI between smoker and nonsmoker
#t_stat is also very closer to zero thus the difference between these two groups is very small 

# now, question is raised, is smoking really the variable that effects BMI? According to this test, no. And in order to improve
# perhaps utilizing Multiple Linear Regression that takes into account all other columns/variables in the dataset
```

#### Nonparametric (chi_square) test

```python
# Research question - Is there association between Sex and Smoker?
# H0: Sex and Smoker have no relationship
# H1: Sex and Smoker they do have relationship 

from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(df['sex'], df['smoker'])

chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print('chi2_test', chi2)
print('p_value', p_value)
print('Degrees of Freedom:', dof)
#print('Expected Frequency', expected)
print('threshold 0.05')

if p_value < 0.05:
    print("There is a statistically significant relationship between Sex and Smoker, thus we reject the null hypothesis")
else:
    print("There is no statistically significant relationship between Sex and Smoker, thus fail to reject the null hypothesis.")
    

#since p_value came less than the set threshold, this tell us that there is relationship between gender and smoker. 
# also looking at the proportions, there is higher proportion of being male smoker than a female

# now, question is raised - how strong is this relationship between sex and smoker. Chi-square test doesn't give us that. It just tells us
# that there is a relationship. To calculate the strength there are other type of tests out there. 
```
